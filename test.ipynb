```python
def get_dict_key(input_dict, value_check):
    for key, value in input_dict.items():
        if value == value_check:
            return key
    return None
```

```python
import os

def list_file_in_folder(path):
    files = [f for f in os.listdir(path) if os.path.isfile(os.path.join(path, f))]
    return files
```

## Mengecek ada label apa di dalam file-file .xml nyah

```python
import xml.etree.ElementTree as ET

def scrap_xml (xml_path):
    labels = []
    tree = ET.parse(xml_path)
    root = tree.getroot()
    for obj in root.findall('object'):
        name = obj.find('name').text
        labels.append(name)
    return labels


xml_location = "voc_labels"

files = list_file_in_folder(xml_location)

all_label = []

for file in files:
    # file = files[i]
    file_path = f"{xml_location}/{file}"
    labels_found = scrap_xml(file_path)
    all_label.extend(labels_found)

all_label = set(all_label)
all_label = list(all_label)

for llabel in all_label:
    print(f"{llabel}")
```

```python
import xml.etree.ElementTree as ET

def scrap_xml (xml_path, label_dict):
    tree = ET.parse(xml_path)
    root = tree.getroot()

    image_filename = root.find('filename').text

    size = root.find('size')
    width = int(size.find('width').text)
    height = int(size.find('height').text)
    depth = int(size.find('depth').text)

    list_label = []

    for obj in root.findall('object'):
        name = obj.find('name').text
        pose = obj.find('pose').text
        # truncated = int(obj.find('truncated').text)
        # difficult = int(obj.find('difficult').text)
        
        if name in label_dict.values():
            index = get_dict_key(label_dict, name)


        bndbox = obj.find('bndbox')
        xmin = int(bndbox.find('xmin').text)
        ymin = int(bndbox.find('ymin').text)
        xmax = int(bndbox.find('xmax').text)
        ymax = int(bndbox.find('ymax').text)

        label = f"{index} {xmin} {ymin} {xmax} {ymax}"
        list_label.append(label)

        # print (label)
    
    return image_filename, list_label
```
